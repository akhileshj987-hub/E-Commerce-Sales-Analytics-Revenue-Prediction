# Convert Order_Date to datetime

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

# Create useful date columns

```python
df["Year"] = df["Order_Date"].dt.year
df["Months"] = df["Order_Date"].dt.month
df["Month_Name"] = df["Order_Date"].dt.month_name()
```

```python
df[["Order_Date", "Year", "Months", "Month_Name"]].head(10)
```

# Now create our most important column: Revenue 💰

The formula is:

**Revenue = Quantity × Unit Price × (1 − Discount)**

```python
df["Revenue"] = (
    df["Quantity"] *
    df["Unit_Price"] *
    (1 - df["Discount"])
)
```

```python
df[["Quantity", "Unit_Price", "Discount", "Revenue"]].head()
```

# ML

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df.columns
```

```python
X = df[["Quantity", "Unit_Price", "Discount"]]

y = df["Revenue"]
```

```python
print("Features (X):")
print(X.head())

print("\nTarget (y):")
print(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.33,
    random_state=42
)
```

```python
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training target:", y_train.shape)
print("Testing target", y_test.shape)
```

# Train Your First ML Model

```python
model = LinearRegression()

model.fit(X_train, y_train)
```

# Make Predictions Using Your Model

```python
y_pred = model.predict(X_test)
```

```python
comparison= pd.DataFrame({
    "Actual Revenue": y_test.values, 
    "Predicted Revenue": y_pred
})

comparison.head()
```

**calculate the model's performance**

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"Mean Absolute Error (MAE): ₹{mae:.2f}")
print(f"Root Mean Squared Error (RMSE): ₹{rmse:.2f}")
print(f"R² Score: {r2:.4f}")
```

**Run this verification code**

```python
# Recalculate predictions
y_pred = model.predict(X_test)

# Recalculate evaluation metrics
mae_check = mean_absolute_error(y_test, y_pred)
mse_check = mean_squared_error(y_test, y_pred)
rmse_check = np.sqrt(mse_check)
r2_check = r2_score(y_test,y_pred)

print("MAE:", mae_check)
print("MSE:", mse_check)
print("RMSE:", rmse_check)
print("R2 Score:", r2_check)

# Verify the relationship between MAE and RMSE
print("\nIs RMSE >= MAE?", rmse_check >= mae_check)
```

# ML model — Random Forest 🌳

```python
from sklearn.ensemble import RandomForestRegressor

print("Random Forest imported successfully!")
```

**Train the Random Forest model**

```python
# Create the Random Forest model 
rf_model = RandomForestRegressor(
    n_estimators=100, 
    random_state=42
)

# Train the model 
rf_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

**Make predictions**

```python
# make prediction using Random Forest
rf_pred = rf_model.predict(X_test)

#compare actual and predicted revenue
rf_comparison = pd.DataFrame({
    "Actual Revenue": y_test.values, 
    "Predctied Revenue": rf_pred
})

rf_comparison.head(10)
```

**Evaluate Random Forest**

```python
# Calculate evaluation metrics
rf_mae = mean_absolute_error(y_test, rf_pred)

rf_mse = mean_squared_error(y_test, rf_pred)

rf_rmse = np.sqrt(rf_mse)

rf_r2 = r2_score(y_test, rf_pred)

print("Random Forest MOdel Performance")
print("-------------------------------")
print(f"MAE:   ₹{rf_mae:.2f}")
print(f"MSE:    {rf_mse:.2f}")
print(f"RMSE:  ₹{rf_rmse:.2f}")
print(f"R² Score:  {rf_r2:.4f}")
```

# final comparison chart

```python
# Compare model performance 
models = ["Linear Regression", "Random Forest"]

mae_score = [mae, rf_mae]
rmse_score =[rmse, rf_rmse]
r2_scores = [r2, rf_r2]

# Plot MAE Comparison 
plt.figure(figsize=(8,5))
plt.bar(models, mae_score)
plt.title("MAE Comparison")
plt.ylabel("Mean Absolute Error (₹)")
plt.tight_layout()
plt.show()

# Plot RMSE Comparison
plt.figure(figsize=(8,5))
plt.bar(models, rmse_score)
plt.title("RMSE Comparison")
plt.ylabel("Root Mean Squared Error (₹)")
plt.tight_layout()
plt.show()

# Plot R² Comparison
plt.figure(figsize=(8,5))
plt.bar(models, r2_scores)
plt.title("R² Score Comparison")
plt.ylabel("R² Score")
plt.ylim(0, 1.05)
plt.tight_layout()
plt.show()
```
